### Movie ticket pricing 
Write a python code that calcluates the ticket price for a movie theatre based on customer age and student status rules are
* Children Under 5 : $0
* Student with ID : $10
* Standard customer : $14
* Senior citizen (65 and above ): $8

```python
name = input("Enter your Name: ")
age = int(input("Enter your Age: "))
is_stud = int(input("""Are You a Student: 
1. Yes
2. No: """))
if is_stud == 1:
    id_card = input("Do You have ID Card (Yes Or No): ")
    if str.lower(id_card) == "yes":
        print("Your Ticket Price is $10")
    elif str.lower(id_card) == 2:
        print("Your Ticket Price is $14")
    else:
        print("Invalid Entry, Enter only yes or no")
elif is_stud == 2:
    if age < 5:
        print("Free Ticket")
    elif age >= 65:
        print("Your Ticket Price is $8")
    else:
        print("Your Ticket Price is $14")
else:
    print("Invalid entry, Choose 1 or 2 only")
```
